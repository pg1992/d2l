# Reading Abalone Data

## Reading the raw data
This is the process for reading the raw data from the zip file downloaded at [UCI ML Archive](https://archive.ics.uci.edu/static/public/1/abalone.zip).

```python
import os
import numpy as np
import pandas as pd
import zipfile

import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

FILENAME = 'abalone.zip'
data_file = os.path.join('data', FILENAME)

with zipfile.ZipFile(data_file) as myzip:
    print(f'{FILENAME} contents:', myzip.namelist())

    # The column names were taken from abalone.names manually.
    names = [
        'Sex',
    	'Length',
    	'Diameter',
    	'Height',
    	'Whole_weight',
    	'Shucked_weight',
    	'Viscera_weight',
    	'Shell_weight',
    	'Rings',
    ]

    with myzip.open('abalone.data') as myfile:
        data = pd.read_csv(myfile, names=names, header=None)
```
